```python
def rabin_karp(text, pattern):
    n = len(text)
    m = len(pattern)
    q = 101  # A prime number
    d = 256  # Number of characters in the input alphabet
    h_pattern = 0  # Hash value for pattern
    h_text = 0     # Hash value for text window
    h = 1
    occurrences = []

    # The value of h would be 'pow(d, m-1)%q'
    for i in range(m - 1):
        h = (h * d) % q

    # Calculate the hash value of pattern and first window of text
    for i in range(m):
        h_pattern = (d * h_pattern + ord(pattern[i])) % q
        h_text = (d * h_text + ord(text[i])) % q

    # Slide the pattern over the text one by one
    for i in range(n - m + 1):
        # If the hash values match then only check for characters one by one
        if h_pattern == h_text:
            # Check for characters one by one
            match = True
            for j in range(m):
                if text[i + j] != pattern[j]:
                    match = False
                    break
            if match:
                occurrences.append(i)

        # Calculate hash value for next window of text: Remove leading digit,
        # add trailing digit
        if i < n - m:
            h_text = (d * (h_text - ord(text[i]) * h) + ord(text[i + m])) % q

            # We might get negative value of t, converting it to positive
            if h_text < 0:
                h_text = (h_text + q)

    return occurrences

# Given inputs
text = "abcababcd"
pattern = "aba"

# Find occurrences
result = rabin_karp(text, pattern)

# Print the results
if result:
    print(f"Pattern '{pattern}' found at indices: {result} in text '{text}'")
else:
    print(f"Pattern '{pattern}' not found in text '{text}'")
```
